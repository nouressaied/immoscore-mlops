```python

import pandas as pd
import numpy as np
import sklearn
import pickle

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
import pandas as pd

df = pd.read_csv(
    "/kaggle/input/competitions/house-prices-advanced-regression-techniques/train.csv"
)

print("Dimensions :", df.shape)
display(df.head())
```

```python

# Sélectionner uniquement les colonnes numériques
df_numeric = df.select_dtypes(include=["number"])

print("Avant nettoyage :", df_numeric.shape)

# Supprimer les lignes contenant des valeurs manquantes
df_clean = df_numeric.dropna()

print("Après nettoyage :", df_clean.shape)

display(df_clean.head())
```

```python

# Variables explicatives
X = df_clean.drop(columns=["SalePrice"])

# Variable cible : prix de vente
y = df_clean["SalePrice"]

print("Dimensions de X :", X.shape)
print("Dimensions de y :", y.shape)
```

```python

from sklearn.model_selection import train_test_split

# Créer 5 catégories de prix
y_bins = pd.qcut(
    y,
    q=5,
    labels=False,
    duplicates="drop"
)

# Découpage 80 % entraînement / 20 % test
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y_bins,
    random_state=42
)

print("X_train :", X_train.shape)
print("X_test :", X_test.shape)
print("y_train :", y_train.shape)
print("y_test :", y_test.shape)
```

```python

from sklearn.linear_model import LinearRegression

# Créer le modèle
model = LinearRegression()

# Entraîner le modèle
model.fit(X_train, y_train)

# Effectuer les prédictions
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)

print("Modèle entraîné avec succès !")
```

```python

from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

# Métriques sur l'entraînement
rmse_train = np.sqrt(mean_squared_error(y_train, y_train_pred))
mae_train = mean_absolute_error(y_train, y_train_pred)
r2_train = r2_score(y_train, y_train_pred)

# Métriques sur le test
rmse_test = np.sqrt(mean_squared_error(y_test, y_test_pred))
mae_test = mean_absolute_error(y_test, y_test_pred)
r2_test = r2_score(y_test, y_test_pred)

print("=== TRAIN ===")
print(f"RMSE : {rmse_train:.2f}")
print(f"MAE  : {mae_train:.2f}")
print(f"R²   : {r2_train:.4f}")

print("\n=== TEST ===")
print(f"RMSE : {rmse_test:.2f}")
print(f"MAE  : {mae_test:.2f}")
print(f"R²   : {r2_test:.4f}")
```

```python

print("===== BILAN DU MODÈLE =====")

print(f"RMSE Train : {rmse_train:.2f}")
print(f"RMSE Test  : {rmse_test:.2f}")

print(f"MAE Train  : {mae_train:.2f}")
print(f"MAE Test   : {mae_test:.2f}")

print(f"R² Train   : {r2_train:.4f}")
print(f"R² Test    : {r2_test:.4f}")

ecart_r2 = r2_train - r2_test

print(f"\nÉcart R² : {ecart_r2:.4f}")

if ecart_r2 > 0.2:
    print("Attention : écart important de généralisation.")
else:
    print("Écart de généralisation plus limité.")

print("\nVersions des bibliothèques :")
print("Scikit-learn :", sklearn.__version__)
print("Pandas :", pd.__version__)
```
